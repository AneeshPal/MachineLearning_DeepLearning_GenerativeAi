```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
# Realational Plotting in Seaborn:

#sns.set_theme(style="ticks")

sns.relplot( # by default relplot creates a scatter plot
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
# Line Plot :

sns.relplot( 
    data=tips,
    x="total_bill",
    y="tip",
    kind="line",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
# Creating different types of plots using Seaborn :

# 1.Scatter Plot:

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
# 2.Line Plot:
# Here we see gray portion along the line plot i.e. confidence interval (ci) .By default ci = 95%
# so shading show mean+ci & mean-ci

flights=sns.load_dataset("flights")
sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
# Categorical Plots:
# 1.Bar Plots
# 2.Box Plots
```

```python
# Bar Plot:

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)

# If we had to do same thing using matplotlib be had to set width for plotting multiple dataset i.e. we had control minute details
```

```python
# Box Plot:


sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# Distribution Plot:
# 1.Histogram

penguins=sns.load_dataset("penguins")

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=10
)
```

```python
# Matrix Plots:
# 1.Heatmaps: Plot rectangular data as color-encoded matrix

flights_pivot=flights.pivot(index="month",columns="year",values="passengers")

sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("Passengers Heatmap")
plt.tight_layout()
```

```python
# Best Practices while using Seaborn :

# 1.Use Seaborn for individual plotting while to add some customisation to it use matplotlib for minute controlling
# i.e. using the best of both the worlds

fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
    errorbar=None # to remove ci
)

ax.set_title("Avg bill value for each year")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
